```python
import os, math, random
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
from PIL import Image

try:
    import cv2
    CV2_OK = True
except Exception as e:
    CV2_OK = False
    print("[warn] cv2 not available:", repr(e))

try:
    import imagehash
    HASH_OK = True
except Exception as e:
    HASH_OK = False
    print("[warn] imagehash not available:", repr(e))

try:
    import torch
    from transformers import AutoImageProcessor, AutoModel
    TORCH_OK = True
except Exception as e:
    TORCH_OK = False
    print("[warn] torch/transformers not available:", repr(e))

try:
    from sklearn.model_selection import StratifiedGroupKFold
    from sklearn.neighbors import NearestNeighbors
    SK_OK = True
except Exception as e:
    SK_OK = False
    print("[warn] scikit-learn not available:", repr(e))

from IPython.display import display

pd.set_option("display.max_columns", 200)

# ============================================================
# 0) Config
# ============================================================

BASE = "/content/drive/MyDrive/kaggle/csiro-biomass/data"
TRAIN_CSV = os.path.join(BASE, "train.csv")
TEST_CSV  = os.path.join(BASE, "test.csv")   # optional
IMG_ROOT  = BASE                             # image_path is relative to here

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

W_MAP = {
    "Dry_Green_g":  0.1,
    "Dry_Dead_g":   0.1,
    "Dry_Clover_g": 0.1,
    "GDM_g":        0.2,
    "Dry_Total_g":  0.5,
}
TARGETS = list(W_MAP.keys())

# toggles (turn off what you don't need)
DO_IMAGE_STATS   = True   # needs cv2
DO_NEAR_DUPES    = True   # needs imagehash
DO_EMBEDDINGS    = True   # needs torch + transformers + sklearn
DO_FOLDS         = True   # needs sklearn

# ============================================================
# 1) Small helpers
# ============================================================

def resolve_image_path(rel_path: str, root=IMG_ROOT):
    """Best-effort absolute path resolution for image_path."""
    if rel_path is None or (isinstance(rel_path, float) and np.isnan(rel_path)):
        return None
    rel = str(rel_path).lstrip("/")
    p = os.path.join(root, rel)
    return p if os.path.exists(p) else None

def show_image_grid(image_paths, title, n=16, cols=4):
    paths = [p for p in image_paths if isinstance(p, str)]
    paths = paths[:n]
    if not paths:
        print("No images for:", title)
        return

    rows = math.ceil(len(paths) / cols)
    plt.figure(figsize=(4*cols, 3.2*rows))
    for i, rel in enumerate(paths):
        ap = resolve_image_path(rel)
        ax = plt.subplot(rows, cols, i+1)
        ax.axis("off")
        ax.set_title(os.path.basename(rel), fontsize=9)

        if ap is None:
            ax.text(0.1, 0.5, "missing", fontsize=10)
            continue

        try:
            if CV2_OK:
                bgr = cv2.imread(ap, cv2.IMREAD_COLOR)
                if bgr is None:
                    ax.text(0.1, 0.5, "read fail", fontsize=10)
                else:
                    rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
                    ax.imshow(rgb)
            else:
                ax.imshow(Image.open(ap).convert("RGB"))
        except Exception as e:
            ax.text(0.1, 0.5, f"err:{type(e).__name__}", fontsize=10)

    plt.suptitle(title)
    plt.tight_layout()
    plt.show()

def plot_hist(x, title, bins=60):
    s = pd.to_numeric(x, errors="coerce").dropna().values
    if len(s) == 0:
        print("No data for:", title)
        return
    plt.figure(figsize=(8,4))
    plt.hist(s, bins=bins)
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.show()

def plot_scatter(x, y, title, xlabel, ylabel, s=10):
    x = pd.to_numeric(x, errors="coerce")
    y = pd.to_numeric(y, errors="coerce")
    m = np.isfinite(x) & np.isfinite(y)
    if m.sum() == 0:
        print("No finite data for:", title)
        return
    plt.figure(figsize=(6,5))
    plt.scatter(x[m], y[m], s=s)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True, alpha=0.3)
    plt.show()

# ============================================================
# 2) Load long train (and test if present)
# ============================================================

assert os.path.exists(TRAIN_CSV), f"Missing: {TRAIN_CSV}"
train = pd.read_csv(TRAIN_CSV)
print("train shape:", train.shape)
print("train columns:", list(train.columns))

test = None
if os.path.exists(TEST_CSV):
    test = pd.read_csv(TEST_CSV)
    print("test shape:", test.shape)
    print("test columns:", list(test.columns))

assert {"image_path","target_name","target"}.issubset(train.columns), "train.csv must have image_path,target_name,target"

# ============================================================
# 3) Metric sanity: global weighted R^2 (raw + log1p) with mean baseline
# ============================================================

def weighted_r2_global(y_true, y_pred, target_names, w_map=W_MAP):
    y_true = np.asarray(y_true, dtype=np.float64)
    y_pred = np.asarray(y_pred, dtype=np.float64)
    tname  = np.asarray(target_names)

    w = np.array([w_map.get(t, 1.0) for t in tname], dtype=np.float64)
    m = np.isfinite(y_true) & np.isfinite(y_pred) & np.isfinite(w)

    y_true = y_true[m]; y_pred = y_pred[m]; w = w[m]
    if len(y_true) == 0 or w.sum() <= 0:
        return np.nan

    ybar = (w * y_true).sum() / w.sum()
    ss_res = (w * (y_true - y_pred)**2).sum()
    ss_tot = (w * (y_true - ybar)**2).sum()
    return np.nan if ss_tot <= 0 else (1.0 - ss_res / ss_tot)

def weighted_r2_global_log1p(y_true, y_pred, target_names, w_map=W_MAP):
    y_true = np.maximum(np.asarray(y_true, dtype=np.float64), 0.0)
    y_pred = np.maximum(np.asarray(y_pred, dtype=np.float64), 0.0)
    return weighted_r2_global(np.log1p(y_true), np.log1p(y_pred), target_names, w_map=w_map)

per_target_mean = train.groupby("target_name")["target"].mean().to_dict()
train["pred_mean"] = train["target_name"].map(per_target_mean).astype(float)

r2_raw = weighted_r2_global(train["target"], train["pred_mean"], train["target_name"])
r2_log = weighted_r2_global_log1p(train["target"], train["pred_mean"], train["target_name"])

print("\n[baseline sanity] per-target mean predictor")
print("global weighted R^2 (raw)  :", r2_raw)
print("global weighted R^2 (log1p):", r2_log)

# ============================================================
# 4) Build wide table (one row per image)
# ============================================================

meta_cols = [c for c in ["image_path","Sampling_Date","State","Species","Pre_GSHH_NDVI","Height_Ave_cm"] if c in train.columns]
wide = (
    train.pivot_table(index=meta_cols, columns="target_name", values="target", aggfunc="first")
         .reset_index()
)

for t in TARGETS:
    if t not in wide.columns:
        wide[t] = np.nan

print("\nwide shape:", wide.shape)
display(wide.head(3))

# ============================================================
# 5) Label QC + regimes + simple flags
# ============================================================

eps = 1e-6

wide["GDM_from_parts"]   = wide["Dry_Green_g"] + wide["Dry_Clover_g"]
wide["Total_from_parts"] = wide["Dry_Green_g"] + wide["Dry_Dead_g"] + wide["Dry_Clover_g"]

wide["GDM_err"]   = wide["GDM_g"] - wide["GDM_from_parts"]
wide["Total_err"] = wide["Dry_Total_g"] - wide["Total_from_parts"]

wide["GDM_relerr"]   = wide["GDM_err"].abs()   / (wide["GDM_g"].abs() + eps)
wide["Total_relerr"] = wide["Total_err"].abs() / (wide["Dry_Total_g"].abs() + eps)

wide["any_negative"] = (
    (wide["Dry_Green_g"]  < -eps) |
    (wide["Dry_Dead_g"]   < -eps) |
    (wide["Dry_Clover_g"] < -eps) |
    (wide["GDM_g"]        < -eps) |
    (wide["Dry_Total_g"]  < -eps)
).astype(int)

wide["component_exceeds_total"] = (
    (wide["Dry_Green_g"]  > wide["Dry_Total_g"] + 1e-3) |
    (wide["Dry_Dead_g"]   > wide["Dry_Total_g"] + 1e-3) |
    (wide["Dry_Clover_g"] > wide["Dry_Total_g"] + 1e-3) |
    (wide["GDM_g"]        > wide["Dry_Total_g"] + 1e-3)
).astype(int)

wide["dead_frac_total"]   = wide["Dry_Dead_g"]   / (wide["Dry_Total_g"] + eps)
wide["clover_frac_total"] = wide["Dry_Clover_g"] / (wide["Dry_Total_g"] + eps)
wide["green_frac_total"]  = wide["Dry_Green_g"]  / (wide["Dry_Total_g"] + eps)
wide["clover_frac_gdm"]   = wide["Dry_Clover_g"] / (wide["GDM_g"] + eps)

wide["clover_only_flag"] = (
    (wide["Dry_Green_g"] <= eps) &
    (wide["Dry_Dead_g"]  <= eps) &
    (wide["Dry_Clover_g"] > 0) &
    (wide["Dry_Total_g"]  > 0)
).astype(int)

wide["dead_dominant_flag"]  = (wide["dead_frac_total"]   >= 0.70).astype(int)
wide["clover_heavy_flag"]   = (wide["clover_frac_total"] >= 0.50).astype(int)
wide["green_dominant_flag"] = (wide["green_frac_total"]  >= 0.70).astype(int)

if "Pre_GSHH_NDVI" in wide.columns:
    wide["ndvi_flag"] = ((wide["Pre_GSHH_NDVI"] < 0.1) | (wide["Pre_GSHH_NDVI"] > 0.9)).astype(int)
else:
    wide["ndvi_flag"] = 0

if "Height_Ave_cm" in wide.columns:
    wide["bio_per_cm"] = wide["Dry_Total_g"] / (pd.to_numeric(wide["Height_Ave_cm"], errors="coerce") + eps)
else:
    wide["bio_per_cm"] = np.nan

print("\n[QC summary]")
print("any_negative             :", int(wide["any_negative"].sum()))
print("component_exceeds_total  :", int(wide["component_exceeds_total"].sum()))
print("GDM_relerr > 1e-4        :", int((wide["GDM_relerr"] > 1e-4).sum()))
print("Total_relerr > 1e-4      :", int((wide["Total_relerr"] > 1e-4).sum()))

print("\n[regimes]")
for c in ["clover_only_flag","clover_heavy_flag","dead_dominant_flag","green_dominant_flag","ndvi_flag"]:
    if c in wide.columns:
        print(f"{c:18s}: {int(wide[c].sum())}")

# show top QC by identity errors
qc_view_cols = meta_cols + [
    "Dry_Green_g","Dry_Dead_g","Dry_Clover_g","GDM_g","Dry_Total_g",
    "GDM_err","GDM_relerr","Total_err","Total_relerr",
    "any_negative","component_exceeds_total"
]
qc_view_cols = [c for c in qc_view_cols if c in wide.columns]

sus_id = wide.sort_values(["Total_relerr","GDM_relerr"], ascending=False)[qc_view_cols]
print("\nTop 20 identity issues:")
display(sus_id.head(20))

# quick plots
for t in ["Dry_Total_g","GDM_g","Dry_Green_g","Dry_Dead_g","Dry_Clover_g"]:
    if t in wide.columns:
        plot_hist(wide[t], f"{t} (raw)")
        plot_hist(np.log1p(np.maximum(pd.to_numeric(wide[t], errors='coerce'), 0)), f"{t} (log1p)")

if "Pre_GSHH_NDVI" in wide.columns:
    plot_scatter(wide["Pre_GSHH_NDVI"], np.log1p(np.maximum(wide["Dry_Total_g"], 0)),
                 "log1p(Total) vs NDVI", "NDVI", "log1p(Dry_Total_g)")

if "Height_Ave_cm" in wide.columns:
    plot_scatter(wide["Height_Ave_cm"], np.log1p(np.maximum(wide["Dry_Total_g"], 0)),
                 "log1p(Total) vs Height", "Height_Ave_cm", "log1p(Dry_Total_g)")

# ============================================================
# 6) Image quality stats (cheap) — optional
# ============================================================

def compute_image_stats_rgb(rgb):
    """
    Cheap stuff that often matters for model robustness:
      - brightness mean/std
      - saturation mean
      - blur var(Laplacian)
      - dark/bright pixel fractions
      - aspect ratio
    """
    out = {}
    if rgb is None:
        return out

    gray = cv2.cvtColor(rgb, cv2.COLOR_RGB2GRAY)
    hsv  = cv2.cvtColor(rgb, cv2.COLOR_RGB2HSV)
    sat  = hsv[...,1]

    out["bright_mean"] = float(gray.mean())
    out["bright_std"]  = float(gray.std())
    out["sat_mean"]    = float(sat.mean())

    lap = cv2.Laplacian(gray, cv2.CV_64F)
    out["blur_var_lap"] = float(lap.var())

    out["dark_frac"]   = float((gray < 25).mean())
    out["bright_frac"] = float((gray > 230).mean())

    h, w = gray.shape
    out["h"] = int(h)
    out["w"] = int(w)
    out["aspect"] = float(w / max(h, 1))
    return out

if DO_IMAGE_STATS:
    if not CV2_OK:
        print("\n[skip] image stats needs cv2.")
    else:
        rows = []
        for rel in tqdm(wide["image_path"].astype(str).tolist(), desc="Image stats"):
            ap = resolve_image_path(rel)
            d = {"image_path": rel, "img_ok": 0}
            if ap is None:
                rows.append(d); continue
            bgr = cv2.imread(ap, cv2.IMREAD_COLOR)
            if bgr is None:
                rows.append(d); continue
            rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
            d["img_ok"] = 1
            d.update(compute_image_stats_rgb(rgb))
            rows.append(d)

        stats_df = pd.DataFrame(rows)
        wide = wide.merge(stats_df, on="image_path", how="left")

        print("\nAdded image stats columns:", [c for c in stats_df.columns if c not in ["image_path","img_ok"]])
        if "bright_mean" in wide.columns:
            plot_scatter(wide["bright_mean"], np.log1p(np.maximum(wide["Dry_Total_g"],0)),
                         "log1p(Total) vs Brightness", "bright_mean", "log1p(Dry_Total_g)")
        if "blur_var_lap" in wide.columns:
            plot_scatter(wide["blur_var_lap"], np.log1p(np.maximum(wide["Dry_Total_g"],0)),
                         "log1p(Total) vs BlurVar(Lap)", "blur_var_lap", "log1p(Dry_Total_g)")

# ============================================================
# 7) Visual review panels (QC / outliers / regimes)
# ============================================================

# A) Identity violations
show_image_grid(
    wide.sort_values(["Total_relerr","GDM_relerr"], ascending=False)["image_path"].tolist(),
    "Top identity violations", n=16, cols=4
)

# B) Biomass per cm outliers
if np.isfinite(pd.to_numeric(wide["bio_per_cm"], errors="coerce")).any():
    show_image_grid(
        wide.sort_values("bio_per_cm", ascending=False)["image_path"].tolist(),
        "Top biomass-per-cm (potential QC)", n=16, cols=4
    )

# C) NDVI flagged
if "ndvi_flag" in wide.columns and int(wide["ndvi_flag"].sum()) > 0:
    show_image_grid(
        wide.loc[wide["ndvi_flag"]==1, "image_path"].tolist(),
        "NDVI outside [0.1, 0.9]", n=16, cols=4
    )

# D) Clover-only regime
if int(wide["clover_only_flag"].sum()) > 0:
    show_image_grid(
        wide.loc[wide["clover_only_flag"]==1, "image_path"].tolist(),
        "Clover-only regime", n=16, cols=4
    )

# ============================================================
# 8) Near-duplicate search (pHash) — optional
# ============================================================

def compute_phash(rel_paths, hash_size=8):
    out = {}
    for rel in tqdm(rel_paths, desc="pHash"):
        ap = resolve_image_path(rel)
        if ap is None:
            out[rel] = None
            continue
        try:
            im = Image.open(ap).convert("RGB")
            out[rel] = imagehash.phash(im, hash_size=hash_size)
        except Exception:
            out[rel] = None
    return out

def find_near_duplicates(hash_dict, max_hamming=6):
    items = [(k,v) for k,v in hash_dict.items() if v is not None]
    pairs = []
    for i in range(len(items)):
        k1, h1 = items[i]
        for j in range(i+1, len(items)):
            k2, h2 = items[j]
            d = (h1 - h2)
            if d <= max_hamming:
                pairs.append((k1, k2, int(d)))
    return pd.DataFrame(pairs, columns=["img1","img2","ham_dist"]).sort_values("ham_dist")

dup_df = pd.DataFrame()
if DO_NEAR_DUPES:
    if not HASH_OK:
        print("\n[skip] near-dupes needs imagehash.")
    else:
        all_imgs = wide["image_path"].astype(str).tolist()
        ph = compute_phash(all_imgs, hash_size=8)
        dup_df = find_near_duplicates(ph, max_hamming=6)
        print("\nNear-duplicate pairs (ham<=6):", len(dup_df))
        display(dup_df.head(30))

        if len(dup_df) > 0:
            # quick inspect a few pairs
            for _, rr in dup_df.head(8).iterrows():
                show_image_grid([rr["img1"], rr["img2"]],
                                f"Near-dupe pair (ham={rr['ham_dist']})",
                                n=2, cols=2)

# ============================================================
# 9) Embedding outliers (DINOv2) — optional
# ============================================================

emb_df = pd.DataFrame()
if DO_EMBEDDINGS:
    if not (TORCH_OK and SK_OK):
        print("\n[skip] embeddings needs torch+transformers and scikit-learn.")
    else:
        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        print("\nEmbedding device:", device)

        MODEL_NAME = "facebook/dinov2-base"
        BATCH = 16

        processor = AutoImageProcessor.from_pretrained(MODEL_NAME)
        model = AutoModel.from_pretrained(MODEL_NAME).to(device).eval()

        @torch.no_grad()
        def embed(rel_paths):
            feats, kept = [], []
            for i in tqdm(range(0, len(rel_paths), BATCH), desc="Embedding"):
                imgs, names = [], []
                for rel in rel_paths[i:i+BATCH]:
                    ap = resolve_image_path(rel)
                    if ap is None:
                        continue
                    try:
                        imgs.append(Image.open(ap).convert("RGB"))
                        names.append(rel)
                    except Exception:
                        continue

                if not imgs:
                    continue

                inp = processor(images=imgs, return_tensors="pt").to(device)
                out = model(**inp)

                # CLS token pooling is stable for ViT backbones
                z = out.last_hidden_state[:, 0]
                z = torch.nn.functional.normalize(z, dim=1)

                feats.append(z.detach().cpu().numpy())
                kept.extend(names)

            if not feats:
                return [], np.zeros((0, 768), dtype=np.float32)

            return kept, np.concatenate(feats, axis=0)

        paths, E = embed(wide["image_path"].astype(str).tolist())
        assert len(paths) == len(E), "Embedding mismatch."

        emb_df = pd.DataFrame({"image_path": paths})
        emb_df = emb_df.merge(wide, on="image_path", how="left")

        k = 10
        nbrs = NearestNeighbors(n_neighbors=min(k+1, len(E)), metric="cosine").fit(E)
        dists, idxs = nbrs.kneighbors(E, return_distance=True)
        emb_df["emb_outlier_score"] = np.median(dists[:, 1:], axis=1)

        print("\nTop embedding outliers:")
        cols = [c for c in ["image_path","emb_outlier_score","State","Sampling_Date","Dry_Total_g","clover_only_flag","clover_heavy_flag"] if c in emb_df.columns]
        display(emb_df.sort_values("emb_outlier_score", ascending=False)[cols].head(20))

        def show_outlier_with_neighbors(row_idx, n_neighbors=7):
            rel = emb_df.iloc[row_idx]["image_path"]
            neigh = idxs[row_idx, 1:1+n_neighbors]
            neigh_paths = [emb_df.iloc[j]["image_path"] for j in neigh]
            show_image_grid([rel] + neigh_paths,
                            f"Outlier + neighbors (score={emb_df.iloc[row_idx]['emb_outlier_score']:.4f})",
                            n=1+n_neighbors, cols=4)

            cols2 = [c for c in ["image_path","Dry_Total_g","GDM_g","Dry_Green_g","Dry_Dead_g","Dry_Clover_g","State","Sampling_Date","Species"] if c in emb_df.columns]
            display(emb_df.loc[[row_idx] + list(neigh), cols2])

        top_idxs = emb_df.sort_values("emb_outlier_score", ascending=False).head(5).index.tolist()
        for ridx in top_idxs:
            show_outlier_with_neighbors(ridx, n_neighbors=7)

# ============================================================
# 10) Safe folds (compact strat + grouped splits) — optional
# ============================================================

if DO_FOLDS:
    if not SK_OK:
        print("\n[skip] folds needs scikit-learn.")
    else:
        assert "Sampling_Date" in wide.columns, "Sampling_Date required to build leakage-aware groups."

        N_SPLITS = 3  # small dataset -> keep folds small to avoid missing regimes

        # clean state/date to avoid weird singleton labels
        state = wide.get("State", pd.Series(["NA"] * len(wide))).astype(str).str.strip().replace({"nan":"NA","None":"NA","": "NA"})
        date  = wide.get("Sampling_Date", pd.Series(["NA"] * len(wide))).astype(str).str.strip().replace({"nan":"NA","None":"NA","": "NA"})

        # bucket ultra-rare states (prevents 1-count issues in folds)
        vc = state.value_counts(dropna=False)
        state_clean = state.where(state.map(vc) >= N_SPLITS, other="Other")

        wide["state_clean"] = state_clean
        wide["date_clean"]  = date

        # group by date×state (reasonable leakage barrier, but not too coarse)
        groups = (wide["date_clean"] + "__" + wide["state_clean"]).values

        # strat label: 4 quantile bins of log1p_total + regime code (compact!)
        wide["log1p_total"] = np.log1p(np.maximum(pd.to_numeric(wide["Dry_Total_g"], errors="coerce").fillna(0), 0))
        nbins = 4
        bins = pd.qcut(wide["log1p_total"].rank(method="first"), q=nbins, labels=False, duplicates="drop").astype(int)
        regime = (wide["clover_only_flag"].astype(int) * 2 + wide["clover_heavy_flag"].astype(int)).astype(int)
        strat = (bins.astype(str) + "_r" + regime.astype(str)).values

        def balance_score(df):
            """Lower is better: penalize missing states and regime imbalance."""
            tab = pd.crosstab(df["fold"], df["state_clean"], normalize="index")
            missing = float((tab == 0.0).sum().sum()) * 10.0 if tab.shape[1] else 0.0
            state_imb = float((tab.max(axis=0) - tab.min(axis=0)).sum()) if tab.shape[1] else 0.0
            co = df.groupby("fold")["clover_only_flag"].mean()
            ch = df.groupby("fold")["clover_heavy_flag"].mean()
            return missing + state_imb + 2.0 * float(co.max() - co.min()) + 1.0 * float(ch.max() - ch.min())

        best = None
        best_seed = None
        best_score = None

        # small seed search makes folds less “accidentally bad”
        for seed in range(0, 200):
            skf = StratifiedGroupKFold(n_splits=N_SPLITS, shuffle=True, random_state=seed)
            fold = np.full(len(wide), -1, dtype=int)
            for f, (_, va_idx) in enumerate(skf.split(wide, strat, groups)):
                fold[va_idx] = f

            tmp = wide.copy()
            tmp["fold"] = fold
            sc = balance_score(tmp)

            if best_score is None or sc < best_score:
                best_score = sc
                best_seed = seed
                best = fold.copy()

        wide["fold"] = best
        print(f"\nSelected fold seed={best_seed} | score={best_score:.4f}")
        print("Fold counts:")
        print(wide["fold"].value_counts().sort_index())

        print("\nState distribution per fold (row-normalized):")
        display(pd.crosstab(wide["fold"], wide["state_clean"], normalize="index").round(3))

        print("\nClover-heavy rate per fold:")
        display(wide.groupby("fold")["clover_heavy_flag"].mean())

        print("\nClover-only rate per fold:")
        display(wide.groupby("fold")["clover_only_flag"].mean())

# ============================================================
# 11) Build “suspect list” table + save artifacts
# ============================================================

sus = wide.copy()

sus["qc_score"] = (
    3.0 * np.clip(pd.to_numeric(sus["Total_relerr"], errors="coerce").fillna(0), 0, 10) +
    2.0 * np.clip(pd.to_numeric(sus["GDM_relerr"],   errors="coerce").fillna(0), 0, 10) +
    1.0 * sus.get("component_exceeds_total", 0).astype(float) +
    1.0 * sus.get("any_negative", 0).astype(float) +
    0.5 * sus.get("ndvi_flag", 0).astype(float)
)

if np.isfinite(pd.to_numeric(sus["bio_per_cm"], errors="coerce")).any():
    b = pd.to_numeric(sus["bio_per_cm"], errors="coerce")
    med = np.nanmedian(b)
    mad = np.nanmedian(np.abs(b - med)) + 1e-9
    z = (b - med) / (1.4826 * mad)
    sus["bio_per_cm_robust_z"] = z
    sus["qc_score"] += 0.5 * np.clip(np.nan_to_num(z, nan=0.0), 0, 20)

keep_cols = meta_cols + [
    "Dry_Green_g","Dry_Dead_g","Dry_Clover_g","GDM_g","Dry_Total_g",
    "GDM_err","GDM_relerr","Total_err","Total_relerr",
    "any_negative","component_exceeds_total",
    "clover_only_flag","clover_heavy_flag","dead_dominant_flag","green_dominant_flag",
    "ndvi_flag","bio_per_cm","bio_per_cm_robust_z",
    "bright_mean","bright_std","sat_mean","blur_var_lap","dark_frac","bright_frac","aspect",
    "fold",
    "qc_score"
]
keep_cols = [c for c in keep_cols if c in sus.columns]

sus = sus[keep_cols].sort_values("qc_score", ascending=False)

print("\nTop 50 QC suspects:")
display(sus.head(50))

OUT_DIR = os.path.join(BASE, "eda_artifacts_clean")
os.makedirs(OUT_DIR, exist_ok=True)

sus_path  = os.path.join(OUT_DIR, "eda_qc_suspects.csv")
wide_path = os.path.join(OUT_DIR, "wide_features.csv")

sus.to_csv(sus_path, index=False)
wide.to_csv(wide_path, index=False)

print("\nSaved:")
print(" -", sus_path)
print(" -", wide_path)

if DO_NEAR_DUPES and len(dup_df) > 0:
    dup_path = os.path.join(OUT_DIR, "near_duplicates_phash.csv")
    dup_df.to_csv(dup_path, index=False)
    print(" -", dup_path)

if DO_EMBEDDINGS and len(emb_df) > 0:
    emb_path = os.path.join(OUT_DIR, "embedding_outliers.csv")
    emb_df.to_csv(emb_path, index=False)
    print(" -", emb_path)

print("\nDone.")
```
